```python
from matplotlib import pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
system_name = 'Pendulum 2D'
dataset_size = [100, 500, 1000]

# Non-manifold (FM and CLF)
fm_separatrix_perc = [0.34, 0.027, 0.022]
fm_f1_scores = [0.9179, 0.9858, 0.9948]

clf_separatrix_perc = [0.0581, 0.0215, 0.0176]
clf_f1_scores = [0.9015, 0.9685, 0.9857]

# Manifold (FM only)
fm_manifold_separatrix_perc = [0.3104, 0.041, 0.02]
fm_manifold_f1_scores = [0.8189, 0.9881, 0.997]

import matplotlib.ticker as mtick

fig, axs = plt.subplots(1, 2, figsize=(13, 5))

# Increase bar spacing by using a wider x range for gaps
x = np.arange(len(dataset_size)) * 2  # Spacing out the bars

width = 0.22  # Slightly wider bar so annotation fits
bar_offsets = [-width * 1.1, 0, width * 1.1]
bar_labels = ['FM (Manifold)', 'FM (No Manifold)', 'CLF (No Manifold)']
bar_colors = ['tab:green', 'tab:blue', 'tab:orange']

# For all data arrays (align ordering for bars and their values)
separatrix_bars = [
    fm_manifold_separatrix_perc,
    fm_separatrix_perc,
    clf_separatrix_perc,
]

# Bar plot for Separatrix Percentages
for bars, label, color, offset in zip(separatrix_bars, bar_labels, bar_colors, bar_offsets):
    axs[0].bar(x + offset, bars, width=width, label=label, color=color)

axs[0].set_xticks(x)
axs[0].set_xticklabels(dataset_size)
axs[0].set_xlabel('Dataset Size')
axs[0].set_ylabel('Separatrix Percentage')
axs[0].yaxis.set_major_formatter(mtick.PercentFormatter(1.0))
axs[0].set_title('Separatrix Percentage')
axs[0].legend()
axs[0].grid(axis='y', linestyle='--', alpha=0.5)

# Line plot for F1 Scores (all solid lines, color is enough for distinction)
axs[1].plot(x, fm_manifold_f1_scores, marker='s', label='FM (Manifold)', color='tab:green', linestyle='-')
axs[1].plot(x, fm_f1_scores, marker='o', label='FM (No Manifold)', color='tab:blue', linestyle='-')
axs[1].plot(x, clf_f1_scores, marker='o', label='CLF (No Manifold)', color='tab:orange', linestyle='-')

axs[1].set_xticks(x)
axs[1].set_xticklabels(dataset_size)
axs[1].set_xlabel('Dataset Size')
axs[1].set_ylabel('F1 Score')
axs[1].set_title('F1 Score')
axs[1].set_ylim(0.8, 1.03)
axs[1].grid(True, linestyle='--', alpha=0.5)
axs[1].legend()

plt.suptitle(f'Performance Metrics by Dataset Size\n{system_name}', fontsize=14)
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```
